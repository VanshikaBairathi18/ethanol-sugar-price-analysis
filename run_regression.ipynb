```python
import pandas as pd
import statsmodels.api as sm
df = pd.read_csv('data/processed/master_data.csv')
df
```

```python
df.dropna(how='all', inplace=True)
df = df.drop(columns=['Unnamed: 5','Unnamed: 6'])

df
```

```python
df_train = df[df['Year'] <= 2025]

target = 'avg_sugar_price'
predictors = ['diversion_pct', 'sugar_production_lmt', 'global_price_inr_per_quintal']

X_train = sm.add_constant(df_train[predictors])
y_train = df_train[target]
model_train = sm.OLS(y_train, X_train).fit()

row_2026 = df[df['Year'] == 2026]
X_2026 = sm.add_constant(row_2026[predictors], has_constant='add')

predicted_2026 = model_train.predict(X_2026).values[0]
actual_2026 = row_2026[target].values[0]

print(f"Predicted 2026 price: {predicted_2026:.2f}")
print(f"Actual 2026 price: {actual_2026:.2f}")
print(f"Gap: {actual_2026 - predicted_2026:.2f} ({(actual_2026-predicted_2026)/actual_2026*100:.1f}%)")
```
